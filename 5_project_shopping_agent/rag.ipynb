```python
from langchain_core.documents import Document

def build_documents(rows):
    documents = []
    for row in rows:
        product_id, name, category, price, description, is_organic = row

        # This is the text that actually gets embedded — what Chroma compares for "meaning"
        text = f"{name}. Category: {category}. {description}"

        # Metadata rides along for filtering — NOT used for semantic matching itself
        metadata = {
            "product_id": product_id,
            "name": name,
            "category": category,
            "price": price,
            "is_organic": bool(is_organic),
        }
        documents.append(Document(page_content=text, metadata=metadata))
    return documents
```
